```python
import torch
import torch.nn as nn

class Model(nn.Module):
  def __init__(self,num_feature):
    super().__init__()
    self.linear=nn.Linear(num_feature,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear(features)
    out=self.sigmoid(out)
    return out
```

```python
features=torch.rand(10,5)
model=Model(features.shape[1])
model(features)
```

```python
print(model.linear.weight)
model.linear.bias
```

```python
#New neural network thoda complex
class Model1(nn.Module):
  def __init__(self,num_feature):
    super().__init__()
    self.linear1=nn.Linear(num_feature,3)
    self.relu=nn.ReLU()
    self.linear2=nn.Linear(3,1)
    self.sigmoid=nn.Sigmoid()

  def forward(self,features):
    out=self.linear1(features)
    out=self.relu(out)
    out=self.linear2(out)
    out=self.sigmoid(out)

    return out

```

```python
features=torch.rand(10,5)
model=Model1(features.shape[1])
model(features)
```

```python
model.linear1.weight
model.linear1.bias
model.linear2.weight
model.linear2.bias
```

```python
import torch.nn as nn

# containers
class Model2(nn.Module):
  def __init__(self,num_feature):
    super().__init__()
    self.network=nn.Sequential(
        nn.Linear(num_feature,3),
        nn.ReLU(),
        nn.Linear(3,1),
        nn.Sigmoid(),
    )

  def forward(self,features):
    out=self.network(features)

    return out
```

```python
features=torch.rand(10,5)
model=Model2(features.shape[1])
ypred=model(features)
```

```python
loss_function=nn.BCELoss()
y_actual=torch.rand(10,1)
loss_function(ypred,y_actual)
```

```python
# torch .optim use
optimizer=torch.optim.SGD(Model2.parameters(),lr=0.1)
optimizer.step()
optimizer.zero_grad()
```
