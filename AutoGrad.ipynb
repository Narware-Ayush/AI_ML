```python
# Design single perceptron from scratch
x=torch.tensor(7.8)
y=torch.tensor(0.0)
w=torch.tensor(1.0)
b=torch.tensor(0.0)
```

```python
def binary_Cross_entropy_loss(prediction,target):
  epsilon=1e-8
  prediction=torch.clamp(prediction,epsilon,1-epsilon)
  return -(target*torch.log(prediction)+(1-target)*torch.log(1-prediction))
```

```python
z=w*x+b
ypred=torch.sigmoid(z)
loss=binary_Cross_entropy_loss(ypred,y)
print(loss);
```

```python
#Derivatives
dloss_dy_pred=(ypred-y)/(ypred*(1-ypred))

dy_pred_dz=ypred*(1-ypred)

dz_dw=x
dz_db=1
dl_dw=dloss_dy_pred*dy_pred_dz*dz_dw
dl_db=dloss_dy_pred*dy_pred_dz*dz_db
```

```python
print(f"manual gradiant of loss by weight :{dl_dw}")
print(f"manual gradian to loss by bias {dl_db}")
```

```python
# using torch liabrary understand single perceptron
w=torch.tensor(1.0,requires_grad=True)
b=torch.tensor(0.0,requires_grad=True)
z=w*x+b
ypred=torch.sigmoid(z)
print(ypred)
```

```python
loss=binary_Cross_entropy_loss(ypred,y)
loss
```

```python
loss.backward()

print(w.grad)
print(b.grad)
```
