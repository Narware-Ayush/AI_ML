```python
x=torch.tensor([[1,2,3,4],[4,5,6,7]])
print(x.shape)
print(x.dtype)
y=torch.empty_like(x)
print(y)
print(torch,torch.zeros_like(x))
print(torch.tensor([[1,2.2323,3,4],[3,4,5,6]], dtype=torch.float32))
```

```python
#methemetical operations on tensors
x=torch.tensor([[2,4,6,6],[3,2,5,8],[9,6,4,11]])
y=torch.tensor([[2,3,4,4],[4,22,7,9],[8,5,2,1]])
print(x+y)
print(x*y)
print(x//2) #int division
print(x/2)
```
