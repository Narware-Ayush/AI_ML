```python
import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
```

```python
df=pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
print(df.head())
```

```python
df.shape
```

```python
xtrain,xtest,ytrain,ytest=train_test_split(df.iloc[:,2:],df.iloc[:,1],test_size=0.2)
df.info()
#print(xtest.head())
print(ytrain.head())
```

```python
scaler=StandardScaler()
xtrain=scaler.fit_transform(xtrain)
xtest=scaler.transform(xtest)
print(xtrain)
```

```python
encoder=LabelEncoder()
ytrain=encoder.fit_transform(ytrain)
ytest=encoder.transform(ytest)
print(ytrain)
print(ytest)
```

```python
xtrain_tensor=torch.from_numpy(xtrain)
xtest_tensor=torch.from_numpy(xtest)
ytrain_tensor=torch.from_numpy(ytrain)
ytest_tensor=torch.from_numpy(ytest)

print(ytest_tensor)
print(xtrain_tensor)
```

```python
class mySimpleNeuNet():

  def __init__(self,x):
    self.weight=torch.rand(x.shape[1],1,requires_grad=True,dtype=torch.float64)
    self.bias=torch.zeros(1,dtype=torch.float64,requires_grad=True)

  def forward(self,x):
    z=torch.matmul(x,self.weight)+self.bias
    print(z)
    ypred=torch.sigmoid(z)
    return ypred

  def loss_function(self,ypred,y):
    epsilon=1e-7
    ypred=torch.clamp(ypred,epsilon,1-epsilon)

    loss=-(ytrain_tensor * torch.log(ypred) + (1-ytrain_tensor) * torch.log(1-ypred)).mean()
    return loss
```

```python
learning_rate=0.1;
epochs=1;
```

```python
#start training pipeline
#create model
model=mySimpleNeuNet(xtrain_tensor)
print(model.weight)
model.bias
```

```python
#define loop
for epoch in range (epochs):
  ypred=model.forward(xtrain_tensor)
  loss=model.loss_function(ypred,ytrain_tensor)
  loss.backward()

  model.weight -= learning_rate * model.weight.grad

  model.bias.grad.zero_()
  model.weight.grad.zero_()

  print(f"print loass at each {epoch}")
```

```python
with torch.no_grad():
  ypred=model.forward(xtest_tensor)
  ypred=(ypred>0.5).float()
  print(ypred)
```
